### Project Objective
The objective of this project is to analyze fictional student spending habits using Python data analysis tools. This notebook processes spending data across different expense categories and payment methods to generate statistical insights and visual charts.

### Dataset Description
The dataset contains fictional records of spending amounts, categories (Food, Transport, Education, Entertainment), and payment methods (Cash, Card, Online).

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

### Library Usage Overview
- **Pandas**: Used to structure data into dataframes, inspect table summaries, filter entries, and group categorical spending.
- **NumPy**: Used to calculate numerical statistics efficiently.
- **Matplotlib**: Used to generate data visualizations like bar charts.

```python
data = {
    "Category": [
        "Food", "Transport", "Education", "Food",
        "Entertainment", "Transport", "Food",
        "Education", "Entertainment", "Transport",
        "Food", "Education"
    ],
    "Amount": [
        500, 150, 800, 350,
        400, 200, 600, 700,
        250, 180, 450, 900
    ],
    "Payment_Method": [
        "Cash", "Card", "Online", "Cash",
        "Card", "Cash", "Online", "Online",
        "Card", "Cash", "Cash", "Online"
    ]
}

df = pd.DataFrame(data)
display(df)
```

### Dataset Structure
The dataset contains three key columns:
- **Category**: The expense classification (e.g., Food, Transport, Education, Entertainment).
- **Amount**: The transaction expense amount in arbitrary units.
- **Payment_Method**: The transaction channel used (Cash, Card, Online).

```python
print("--- First 5 Rows ---")
display(df.head())

print("\n--- Dataset Shape ---")
print(df.shape)

print("\n--- Summary Info ---")
df.info()

print("\n--- Statistical Overview ---")
display(df.describe(include="all"))
```

### Exploration Findings
- **Shape**: The dataset contains 12 rows and 3 columns.
- **Data Types**: `Category` and `Payment_Method` are object/string types, while `Amount` is an integer.
- **Summary**: `Food` is the most frequent category (4 occurrences), and `Cash` is the most frequent payment method (5 occurrences).

```python
print("Missing Values Count:")
print(df.isnull().sum())

print("\nDuplicate Rows Count:", df.duplicated().sum())
```

### Data Quality Assessment
- **Missing Values**: 0 missing entries detected across all columns.
- **Duplicates**: 0 duplicate rows detected.
- **Result**: The dataset is complete, clean, and structured properly for numerical analysis.

```python
# Question 1: Total spending across all entries
total_spending = df["Amount"].sum()
print("Total spending:", total_spending)

# Question 2: Average spending per entry
average_spending = np.mean(df["Amount"])
print("Average spending:", average_spending)

# Question 3: Category with the highest total spending
category_spending = df.groupby("Category")["Amount"].sum()
print("\nSpending by Category:\n", category_spending)
highest_category = category_spending.idxmax()
print("Category with the highest spending:", highest_category)

# Question 4: Highest individual expense
highest_expense = df["Amount"].max()
print("\nHighest individual expense:", highest_expense)

# Question 5: Total spending through each payment method
payment_spending = df.groupby("Payment_Method")["Amount"].sum()
print("\nSpending by Payment Method:\n", payment_spending)
```

### Detailed Statistical Explanations
- **Total Spend**: The cumulative expenditure across all 12 recorded items is **5,430 units**.
- **Average Spend**: The mean expenditure per transaction entry is **452.5 units**.
- **Top Expense Category**: **Education** accumulated the highest total spending at **2,400 units**.
- **Max Single Transaction**: The largest single transaction recorded was **900 units** (under Education).
- **Payment Breakdown**: Online transactions generated the highest overall payment total (**3,000 units**), followed by Cash (**1,780 units**) and Card (**650 units**).

```python
category_spending = df.groupby("Category")["Amount"].sum()

plt.figure(figsize=(8, 5))
category_spending.plot(kind="bar", color="skyblue", edgecolor="black")
plt.title("Total Spending by Category")
plt.xlabel("Expense Category")
plt.ylabel("Total Amount")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

### Visualization Interpretation
The bar chart clearly displays that **Education** comprises the largest expense total, exceeding 2,000 units. **Food** represents the second-largest overall expense group, while **Entertainment** and **Transport** account for smaller portions of the overall student expenditure. The visualization provides a clear relative breakdown of total spending per category at a glance.
